## 3. CONTEO DE FRECUENCIAS (PATRÓN HASH MAP)

```python
accesos_servidor = ["192.168.1.1", "10.0.0.2", "192.168.1.1", "172.16.0.5", "192.168.1.1", "10.0.0.2"]
conteo_ips = {}

for ip in accesos_servidor:
    conteo_ips[ip] = conteo_ips.get(ip, 0) + 1

print("\nFrecuencia de accesos por IP:", conteo_ips)
```

```python
# Creo un diccionario llamado inventario para guardar varios productos usando su SKU como clave.
inventario = {
    # Cada SKU identifica un producto y su valor es otro diccionario con sus datos.
    "SKU101": {"nombre": "Impresora", "precio": 89.99, "existencia": 6},
    "SKU102": {"nombre": "Router", "precio": 34.50, "existencia": 20},
    "SKU103": {"nombre": "Disco Duro", "precio": 55.00, "existencia": 14},
    "SKU104": {"nombre": "Cargador", "precio": 9.99, "existencia": 40}
}

print("--- Inventario ---")
# Recorro cada clave y cada registro guardado dentro del diccionario inventario.
for sku, datos in inventario.items():
    # Imprimo el SKU y los datos del producto con un formato más ordenado.
    # :<10 deja el nombre alineado a la izquierda en un espacio de 10 caracteres.
    # :.2f muestra el precio con exactamente dos decimales.
    print(f"SKU: {sku} | Nombre: {datos['nombre']:<10} "
        f"| Precio: ${datos['precio']:.2f} | Existencia: {datos['existencia']}")


# 2. Contador de ocurrencias: frecuencia de palabras en un texto
# Creo una lista con palabras repetidas para saber cuántas veces aparece cada una.
palabras = ["python", "java", "python", "python", "c++", "java", "python"]

# Creo un diccionario vacío donde guardaré cada palabra y su cantidad de apariciones.
conteo = {}
# Recorro una por una todas las palabras de la lista.
for palabra in palabras:
    # get devuelve el conteo actual o 0 si la palabra todavía no existe.
    # Después sumo 1 para registrar la aparición actual.
    conteo[palabra] = conteo.get(palabra, 0) + 1

print("\nFrecuencia de palabras:", conteo)


# 3. Agrupamiento por categoría
# Creo una lista de tuplas donde cada tupla contiene un alumno y su materia.
alumnos = [
    ("Henry", "Matemática"),
    ("Oscar", "Historia"),
    ("Isvi", "Matemática"),
    ("Gabo", "Ciencias"),
    ("Rommel", "Historia"),
    ("Aldair", "Ciencias")
]

# Creo un diccionario vacío para agrupar los nombres según la materia.
alumnos_por_materia = {}
# Recorro cada tupla y separo el nombre del alumno y la materia.
for nombre, materia in alumnos:
    # Compruebo si la materia todavía no tiene una lista creada.
    if materia not in alumnos_por_materia:
        # Si no existe, creo una lista vacía para guardar sus alumnos.
        alumnos_por_materia[materia] = []
    # Agrego el nombre del alumno a la lista de su materia.
    alumnos_por_materia[materia].append(nombre)

print("\nAlumnos agrupados por materia:")
# Recorro cada materia junto con la lista de alumnos que pertenecen a ella.
for materia, lista in alumnos_por_materia.items():
    # Imprimo cada materia y sus alumnos agrupados.
    print(f"{materia}: {lista}")
```
